```python
import pandas as pd
import numpy as np

# Load the main employee dataset
df = pd.read_csv("employee_attrition_2026_master.csv")

# Load the data dictionary
data_dictionary = pd.read_csv("data_dictionary.csv")

# Display the first 5 rows
display(df.head())
```

```python
# Count missing values in each column
missing_values = df.isnull().sum()

print("Missing values in each column:")
display(missing_values)

# Calculate missing-value percentage
missing_percentage = (df.isnull().sum() / len(df)) * 100

print("Missing value percentage:")
display(missing_percentage.round(2))
```

```python
# Count duplicate rows
duplicate_count = df.duplicated().sum()

print("Total duplicate rows:", duplicate_count)

# Display duplicate records
duplicate_rows = df[df.duplicated()]

display(duplicate_rows.head())
```

```python
# Identify numerical columns
numerical_columns = df.select_dtypes(include=np.number).columns

print("Numerical columns:")
print(numerical_columns.tolist())
```

```python
# Select relevant numerical columns for outlier detection
columns_for_iqr = [
    'age',
    'tenure_years',
    'salary_usd',
    'salary_vs_market_pct',
    'days_in_office_required',
    'commute_minutes',
    'flexibility_importance',
    'weekly_hours',
    'after_hours_work',
    'burnout_score',
    'manager_relationship_score',
    'manager_1on1_per_month',
    'engagement_score',
    'promotions_last_2y',
    'career_growth_score',
    'team_size'
]

# Calculate IQR and identify outliers
outlier_results = []

for col in columns_for_iqr:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = df[
        (df[col] < lower_bound) |
        (df[col] > upper_bound)
    ]

    outlier_results.append({
        'Column': col,
        'Q1': Q1,
        'Q3': Q3,
        'IQR': IQR,
        'Lower Bound': lower_bound,
        'Upper Bound': upper_bound,
        'Outlier Count': len(outliers)
    })

outlier_summary = pd.DataFrame(outlier_results)

display(outlier_summary.round(2))
```

```python
# Inspect columns with potential outliers

outlier_columns = [
    'age',
    'tenure_years',
    'salary_usd',
    'commute_minutes',
    'weekly_hours',
    'engagement_score',
    'manager_1on1_per_month'
]

# Display statistical summary
display(df[outlier_columns].describe().round(2))
```

```python
# Check frequency of manager meetings per month

meeting_distribution = (
    df['manager_1on1_per_month']
    .value_counts()
    .sort_index()
)

print("Manager 1:1 meeting distribution:")
display(meeting_distribution)
```

```python
# Identify categorical columns
categorical_columns = df.select_dtypes(include='object').columns

print("Categorical columns:")
print(categorical_columns.tolist())
```

```python
categorical_columns = df.select_dtypes(include='str').columns

print("Categorical columns:")
print(categorical_columns.tolist())
```

```python
# Inspect unique values in categorical columns

for col in categorical_columns:
    print(f"\nColumn: {col}")
    print(df[col].value_counts(dropna=False).head(15))
```

```python
# Exclude employee_id because it contains unique identifiers
text_columns = [
    col for col in categorical_columns
    if col != 'employee_id'
]

# Display unique values in each text column
for col in text_columns:
    print(f"\n--- {col} ---")

    for value in sorted(df[col].dropna().unique()):
        print(repr(value))
```

```python
# Check leading/trailing spaces and casing inconsistencies

for col in text_columns:
    values = df[col].dropna().unique()

    # Find values with extra spaces
    extra_spaces = [
        value for value in values
        if value != value.strip()
    ]

    # Find different spellings/casing of the same text
    normalized = {}

    for value in values:
        key = value.strip().casefold()
        normalized.setdefault(key, []).append(value)

    inconsistent_values = {
        key: original_values
        for key, original_values in normalized.items()
        if len(original_values) > 1
    }

    print(f"\nColumn: {col}")
    print("Extra spaces:", extra_spaces)
    print("Casing variations:", inconsistent_values)
```

```python
# Create a copy of the original dataset
df_clean = df.copy()

# Remove leading and trailing spaces from text columns
for col in text_columns:
    df_clean[col] = df_clean[col].str.strip()

print("String standardisation completed.")

# Verify the result
for col in text_columns:
    print(col, ":", df_clean[col].nunique(), "unique values")
```

```python
# Display the data dictionary
display(data_dictionary.head(20))
```

```python
# Inspect maximum and minimum values
for col in outlier_columns:
    print(f"\n{col}")
    print("Minimum:", df_clean[col].min())
    print("Maximum:", df_clean[col].max())
    print("Unique values:", sorted(df_clean[col].unique())[-10:])
```

```python
# Inspect remote employees with zero office days
remote_check = df_clean[
    (df_clean['work_arrangement'] == 'remote') &
    (df_clean['days_in_office_required'] == 0) &
    (df_clean['commute_minutes'] == 180)
]

display(remote_check[
    ['employee_id', 'work_arrangement',
     'days_in_office_required', 'commute_minutes']
])
```

```python
# Calculate IQR bounds for weekly working hours
Q1 = df_clean['weekly_hours'].quantile(0.25)
Q3 = df_clean['weekly_hours'].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)

# Inspect potential outliers
weekly_outliers = df_clean[
    (df_clean['weekly_hours'] < lower_bound) |
    (df_clean['weekly_hours'] > upper_bound)
]

print("Total outliers:", len(weekly_outliers))

display(weekly_outliers[
    ['employee_id', 'job_level', 'weekly_hours']
].head(20))
```

```python
# Cap weekly working hours using IQR upper bound
df_clean['weekly_hours'] = df_clean['weekly_hours'].clip(
    lower=lower_bound,
    upper=upper_bound
)

print("Outlier capping completed.")

print("Maximum weekly hours after capping:",
      df_clean['weekly_hours'].max())

print("Values above upper bound:",
      (df_clean['weekly_hours'] > upper_bound).sum())
```
